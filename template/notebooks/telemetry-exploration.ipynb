```python
%load_ext autoreload
%autoreload 2

import sys
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

sys.path.insert(0, str(Path.cwd()))
import frontend_model as fm

SEED = 42
rng = np.random.default_rng(SEED)
```

```python
sensor = fm.Sensor(time_constant_s=0.8)
readings = [sensor.step(600, 0.05, rng.normal(0, 5)) for _ in range(200)]
first = fm.crosses_threshold(readings, 300)

fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(readings, label="reading")
ax.axhline(300, color="tab:red", linestyle="--", label="threshold")
if first is not None:
    ax.axvline(first, color="tab:gray", linestyle=":", label="first crossing")
ax.set_xlabel("sample (50 ms)")
ax.set_ylabel("ADC counts")
ax.legend();
```
